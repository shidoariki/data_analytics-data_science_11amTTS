### Step 0: Environment Setup & SQLite Connection

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

# Connect to local database
conn = sqlite3.connect('session_22_analytics.db')
```

### Task 1: Load Restaurants Table into DataFrame via `read_sql()` and Display First 5 Rows

```python
df_restaurants = pd.read_sql("SELECT name, cuisine, rating FROM restaurants", conn)
df_restaurants.head()
```

### Task 2: Jupyter Notebook `%sql` Magic Command - Movies Rated > 8.0

```python
# In Jupyter: 
# %load_ext sql
# %sql sqlite:///session_22_analytics.db
# %%sql
query_task2 = "SELECT movie_id, title, genre, rating FROM movies WHERE rating > 8.0 ORDER BY rating DESC"
pd.read_sql(query_task2, conn)
```

### Task 3: Load Last 100 Orders & Find Top 3 Most Ordered Food Items

```python
# Load last 100 orders using SQL LIMIT
df_orders_100 = pd.read_sql("SELECT * FROM orders ORDER BY order_id DESC LIMIT 100", conn)

# Find top 3 most ordered items using DataFrame value_counts()
top_3_items = df_orders_100['item_name'].value_counts().head(3)
print("Top 3 Most Ordered Food Items:")
print(top_3_items)
```

### Task 4: Load IPL Matches and Plot Matches Played per Team (Matplotlib)

```python
df_matches = pd.read_sql("SELECT team1, team2 FROM matches", conn)
all_teams = pd.concat([df_matches['team1'], df_matches['team2']])
team_counts = all_teams.value_counts()

plt.figure(figsize=(10, 5))
team_counts.plot(kind='bar', color='#2563EB', edgecolor='#1E3A8A')
plt.title('IPL 2023: Total Matches Played by Each Team', fontsize=14, fontweight='bold')
plt.xlabel('Team')
plt.ylabel('Matches Played')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Task 5: AI-Assisted Query - Users Who Placed More Than 5 Orders

```python
ai_query = """
SELECT 
    u.user_id,
    u.user_name,
    u.city,
    COUNT(o.order_id) AS total_orders,
    ROUND(SUM(o.amount), 2) AS total_spend
FROM users u
INNER JOIN orders o ON u.user_id = o.user_id
GROUP BY u.user_id, u.user_name, u.city
HAVING COUNT(o.order_id) > 5
ORDER BY total_orders DESC;
"""
df_loyal_users = pd.read_sql(ai_query, conn)
df_loyal_users
```
